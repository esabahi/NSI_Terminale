## 1. Rappel sur la programmation fonctionnelle
La **programmation fonctionnelle** consiste à construire des programmes en combinant des fonctions.

Concepts clés :
- Les fonctions sont des objets de première classe
- On évite les effets de bord
- Fonctions d'ordre supérieur : prennent ou retournent des fonctions

## 2. Rappel sur la syntaxe Python des fonctions
- Fonction classique :
```python
def nom_fonction(param1, param2):
    return resultat
```
- Lambda (anonyme) : `lambda x: x**2`
- Fonction d'ordre supérieur :
```python
def appliquer(f, lst):
    return [f(x) for x in lst]
```

## 3. Les fonctions d'ordre supérieur

Les fonctions d'ordre supérieur sont des fonctions qui peuvent :
- Prendre d'autres fonctions en paramètres
- Retourner des fonctions comme résultat

````python
def appliquer_operation(fonction, valeur):
    return fonction(valeur)

def multiplier_par_deux(x):
    return x * 2

def elever_au_carre(x):
    return x ** 2

print("\n=== FONCTIONS D'ORDRE SUPÉRIEUR ===")
resultat1 = appliquer_operation(multiplier_par_deux, 5)
resultat2 = appliquer_operation(elever_au_carre, 4)
print(f"5 * 2 = {resultat1}")
print(f"4² = {resultat2}")
````

Exemple :

## 8.Synthèse
